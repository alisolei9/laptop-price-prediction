# Data Cleaning

## Objective

Prepare the raw dataset for feature engineering by applying reproducible cleaning steps while keeping the raw dataset unchanged.

```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

```python
DATA_PATH = Path("../data/raw/laptop_price - dataset.csv")

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
df_clean = df.copy()

df_clean.head()
```

```python
df_clean.shape
```

```python
df_clean.info()
```

```python
missing_values = df_clean.isnull().sum()

missing_values
```

```python
missing_report = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Missing Percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
})

missing_report
```

```python
duplicate_count = df_clean.duplicated().sum()

duplicate_count
```

```python
duplicate_percentage = round(
    duplicate_count / len(df_clean) * 100,
    2
)

print(f"Duplicate Rows: {duplicate_count}")
print(f"Duplicate Percentage: {duplicate_percentage}%")
```

```python
column_mapping = {
    "Company": "company",
    "Product": "product",
    "TypeName": "type_name",
    "Inches": "inches",
    "ScreenResolution": "screen_resolution",
    "CPU_Company": "cpu_company",
    "CPU_Type": "cpu_type",
    "CPU_Frequency (GHz)": "cpu_frequency",
    "RAM (GB)": "ram",
    "Memory": "memory",
    "GPU_Company": "gpu_company",
    "GPU_Type": "gpu_type",
    "OpSys": "operating_system",
    "Weight (kg)": "weight",
    "Price (Euro)": "price"
}
```

```python
column_mapping
```

```python
df_clean = df_clean.rename(columns=column_mapping)

df_clean.head()
```

```python
df_clean.columns
```

```python
df_clean.shape
```

```python
df_clean.info()
```

### Column Standardization Summary

Column names were standardized using snake_case naming convention.

Special characters, spaces, units and inconsistent naming styles were removed to improve readability and maintainability.

```python
df_clean.dtypes
```

```python
dtype_report = pd.DataFrame({
    "Column": df_clean.columns,
    "Data Type": df_clean.dtypes.values
})

dtype_report
```

```python
print("Shape:", df_clean.shape)
print()

print("Columns:")
print(df_clean.columns.tolist())
```

```python
expected_columns = [
    "company",
    "product",
    "type_name",
    "inches",
    "screen_resolution",
    "cpu_company",
    "cpu_type",
    "cpu_frequency",
    "ram",
    "memory",
    "gpu_company",
    "gpu_type",
    "operating_system",
    "weight",
    "price"
]
```

```python
PROCESSED_DATA_PATH = Path("../data/processed/laptop_cleaned.csv")

df_clean.to_csv(PROCESSED_DATA_PATH, index=False)

print(f"Processed dataset saved to:\n{PROCESSED_DATA_PATH}")
```
