# Model Comparison

Compare all trained machine learning models.

```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame(
    {
        "Model": [
            "Linear Regression",
            "Decision Tree",
            "Random Forest",
            "Random Forest (Tuned)",
        ],

        "MAE": [
            241.02,
            237.66,
            179.93,
            176.52,
        ],

        "RMSE": [
            326.27,
            363.02,
            268.42,
            262.20,
        ],

        "R2": [
            0.7855,
            0.7345,
            0.8548,
            0.8615,
        ],
    }
)

results
```

## Model Comparison (R² Score)

```python
plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["R2"],
)

plt.title("Model Comparison (R²)")
plt.xlabel("Model")
plt.ylabel("R² Score")

plt.xticks(rotation=15)

plt.grid(axis="y")

plt.show()
```

## Model Comparison (MAE)

```python
plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["MAE"],
)

plt.title("Model Comparison (MAE)")
plt.xlabel("Model")
plt.ylabel("MAE")

plt.xticks(rotation=15)

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["RMSE"],
)

plt.title("Model Comparison (RMSE)")
plt.xlabel("Model")
plt.ylabel("RMSE")

plt.xticks(rotation=15)

plt.grid(axis="y")

plt.show()
```

## Best Model

```python
best_model = results.loc[
    results["R2"].idxmax()
]

best_model
```

```python
print(f"Best Model : {best_model['Model']}")
print(f"R² Score   : {best_model['R2']:.4f}")
print(f"MAE        : {best_model['MAE']:.2f}")
print(f"RMSE       : {best_model['RMSE']:.2f}")
```
