# Final Model Comparison

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Model Performance

```python
results = pd.DataFrame(
    {
        "Model": [
            "Linear Regression",
            "Decision Tree",
            "Random Forest",
            "Random Forest Tuned",
            "KNN",
            "SVR",
            "Gradient Boosting",
            "XGBoost",
            "LightGBM",
            "CatBoost",
        ],
        "MAE": [
            241.02,
            237.66,
            179.93,
            176.52,
            226.49,
            919.32,
            185.16,
            181.47,
            180.74,
            174.51,
        ],
        "RMSE": [
            326.27,
            363.02,
            268.42,
            262.20,
            353.64,
            649.83,
            253.31,
            262.39,
            290.55,
            254.08,
        ],
        "R2": [
            0.7855,
            0.7345,
            0.8548,
            0.8615,
            0.7480,
            -0.0296,
            0.8707,
            0.8613,
            0.8299,
            0.8699,
        ],
    }
)
```

```python
results
```

## Sort by R²

```python
results = results.sort_values(
    by="R2",
    ascending=False,
)

results
```

## Best Model

```python
results.head(1)
```

```python
best_model = results.iloc[0]

print("Best Model")
print(best_model)
```

## Visualization

```python
plt.figure(figsize=(10,6))

plt.bar(
    results["Model"],
    results["R2"],
)

plt.xticks(rotation=45)

plt.ylabel("R²")

plt.title("Model Comparison (R²)")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.bar(
    results["Model"],
    results["MAE"],
)

plt.xticks(rotation=45)

plt.ylabel("MAE")

plt.title("Model Comparison (MAE)")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.bar(
    results["Model"],
    results["RMSE"],
)

plt.xticks(rotation=45)

plt.ylabel("RMSE")

plt.title("Model Comparison (RMSE)")

plt.tight_layout()

plt.show()
```

# Final Conclusion

Ten regression models were evaluated for laptop price prediction.

Among all models, Gradient Boosting achieved the highest R² score.

CatBoost showed nearly identical performance and achieved the lowest MAE.

Decision Tree and SVR showed the weakest performance.

Gradient Boosting was selected as the final model because it provided the best overall balance between prediction accuracy and generalization.
