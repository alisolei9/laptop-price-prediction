```python
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

from src.data_loader import load_data
from src.preprocessing import clean_data
from src.feature_engineering import engineer_features
from src.encoding import encode_features
```

```python
#Pipeline
df_raw = load_data("laptop_price - dataset.csv")

df_clean = clean_data(df_raw)

df_featured = engineer_features(df_clean)

df_encoded = encode_features(df_featured)

df_encoded.head()
```

```python
df_encoded.shape
```

## Split Features and Target

Machine learning models learn the relationship between input features (`X`) and the target variable (`y`).

In this project:

- Features (`X`) → Laptop specifications
- Target (`y`) → Laptop price

```python
X = df_encoded.drop(columns="price")

y = df_encoded["price"]
```

```python
print("Features Shape :", X.shape)

print("Target Shape :", y.shape)
```

```python
X.head()
```

```python
y.head()
```

## Train-Test Split

The dataset is divided into two parts:

- Training set (80%)
- Testing set (20%)

The model learns from the training set and is evaluated on the testing set.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train :", X_train.shape)

print("X_test  :", X_test.shape)

print("y_train :", y_train.shape)

print("y_test  :", y_test.shape)
```

```python
X_train.head()
```

```python
y_train.head()
```

## Save Prepared Datasets

The training and testing datasets are saved for use in the following notebooks.

This avoids repeating the preprocessing pipeline every time.

```python
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

DATA_PROCESSED.mkdir(exist_ok=True)
```

```python
X_train.to_csv(DATA_PROCESSED / "X_train.csv", index=False)

X_test.to_csv(DATA_PROCESSED / "X_test.csv", index=False)

y_train.to_csv(DATA_PROCESSED / "y_train.csv", index=False)

y_test.to_csv(DATA_PROCESSED / "y_test.csv", index=False)
```
